```python
import torch
import numpy as np
import pandas as pd
```

```python
# creating a tensor
x = torch.tensor([[1,2,3],[4,5,6]])
print(f'The tensor is {x}')
print('-'*50)
print('-'*50)
print(f'The shape of the tensor is {x.shape}')
```

```python
# Some tensor Operation
a = torch.tensor([1,2,3])
b = torch.tensor([4,5,6])

print(f' The Adddition of the tensor is {a+b}')
print('-'*50)
print('-'*50)
print(f' The Subtraction of the tensor is {a-b}')
```

```python
# Some Matrix Operation
# help(torch.rand) Learning about torch.rand() function

a = torch.rand(2,3)
b = torch.rand(3,2)

print('The matrix A is' , a)
print('_'*50)
print('_'*50)
print('The matrix B is' , b)
print('_'*50)
print('_'*50)
print(f'{torch.matmul(a,b)}')
```

```python
# Tensor Shapes

x = torch.rand(2,3,4)
x.shape  # Two tensors will form of shape (3,4)
print('_'*50)
print(f'The initial shape of the tensor is {x.shape}')
print('_'*50)
print(x)

# Reshaping the tensor

x = x.reshape(6,4)
# x = x.reshape(2,12)
print('_'*50)
print(f'The new shape of the tensor is {x.shape}')
print('_'*50)
print(x)


# The total number of elements in a tensor must be same before and after applying the reshaping.
```
