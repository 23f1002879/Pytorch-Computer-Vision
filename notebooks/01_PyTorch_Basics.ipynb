```python
import torch
import numpy as np
import pandas as pd
```

```python
# creating a tensor
x = torch.tensor([[1,2,3],[4,5,6]])
print(f'The tensor is {x}')
print('-'*50)
print('-'*50)
print(f'The shape of the tensor is {x.shape}')
```

```python
# Some tensor Operation
a = torch.tensor([1,2,3])
b = torch.tensor([4,5,6])

print(f' The Adddition of the tensor is {a+b}')
print('-'*50)
print('-'*50)
print(f' The Subtraction of the tensor is {a-b}')
```

```python
# Some Matrix Operation
# help(torch.rand) Learning about torch.rand() function

a = torch.rand(2,3)
b = torch.rand(3,2)

print('The matrix A is' , a)
print('_'*50)
print('_'*50)
print('The matrix B is' , b)
print('_'*50)
print('_'*50)
print(f'{torch.matmul(a,b)}')
```

```python
# Tensor Shapes

x = torch.rand(2,3,4)
x.shape  # Two tensors will form of shape (3,4)
print('_'*50)
print(f'The initial shape of the tensor is {x.shape}')
print('_'*50)
print(x)

# Reshaping the tensor

x = x.reshape(6,4)
# x = x.reshape(2,12)
print('_'*50)
print(f'The new shape of the tensor is {x.shape}')
print('_'*50)
print(x)


# The total number of elements in a tensor must be same before and after applying the reshaping.
```

```python
# Checking for GPU
device = 'cuda' if torch.cuda.is_available() else 'CPU'
print(device)
```

```python
# The automatic differentiation
x = torch.tensor(3.0, requires_grad=True) # The require_grad method makes the computational graph i.e it will keep a track of every operation which we will perform on that tensor.
print(x)

y = x**3 + 2*x**2 + 10

y.backward() # .backward() function triggers the chain rule of the calculus
print(x.grad)
```

```python
# Setting up a simple linear regression model

import torch.nn as nn

model = nn.Linear(3,1)

x = torch.rand(1,3)
y = torch.rand(1,3)

criterion = nn.MSELoss()
optimizer = torch.optim.SGD(model.parameters(), lr=0.01)
```

```python
for epoch in range(100):

    pred = model(x)

    loss = criterion(pred,y)

    optimizer.zero_grad() # This clears out (resets to zero) all the old gradients stored in the model from the previous loop.

    loss.backward()

    optimizer.step()

print(loss.item())
```
