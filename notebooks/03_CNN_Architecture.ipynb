## Building my first CNN


```python
import torch
import torchvision
import torch.nn as nn
from torchvision import transforms, datasets
from torch.utils.data import DataLoader, Dataset
```

```python
class CNN(nn.Module):   # Inheriting the nn.Module class
  def __init__(self):
    super().__init__()    # Calling the parent constructor

    self.conv1 = nn.Conv2d(
        in_channels=1,
        out_channels=32,
        kernel_size=3,
        padding=1,
    )

    self.relu = nn.ReLU()
    self.pool = nn.MaxPool2d(2)
    self.fc = nn.Linear(
        32*14*14,
        10
    )

  def forward(self, x):
    x = self.conv1(x)
    x = self.relu(x)
    x = self.pool(x)
    x = x.view(x.size(0), -1)
    x = self.fc(x)
    return x
```

```python
model = CNN()
```

```python
# Loading the dataset

transform = transforms.ToTensor()

trainDataset = datasets.MNIST(
    root='data',
    train=True,
    download=True,
    transform=transform
)

testDataset = datasets.MNIST(
    root='data',
    train=False,
    download=True,
    transform=transform
)

print(type(trainDataset))
```

```python
trainLoader = DataLoader(
    trainDataset,
    batch_size=32,
    shuffle=True
)

testLoader = DataLoader(
    testDataset,
    batch_size=32,
    shuffle=False
)
```

```python
out = model(trainLoader)

print(out.shape)
```
